```python
m_array = np.array((13.0, 13.5, 14.0))
z_array = np.array(("000", "050", "103"))

X_train_rad, y_train_qdm, y_train_sdm, y_train_sdm_qdm = get_train(m_array, z_array, islog=True)

mask = X_train_rad[:, 0] >= -1.0
X_train_rad = X_train_rad[mask]
y_train_qdm = y_train_qdm[mask]
y_train_sdm = y_train_sdm[mask]
y_train_sdm_qdm = y_train_sdm_qdm[mask]
```

```python
est_qdm, next_qdm, score_qdm = perform_sr(X_train_rad, y_train_qdm, 
                                       init_depth=(2, 6), 
                                       population_size=50000, generations=50, stopping_criteria=0.01, 
                                       p_crossover=0.6, p_subtree_mutation=0.2, p_hoist_mutation=0.05, p_point_mutation=0.15, 
                                       metric=mae_w_penalty_log, 
                                       function_set=('add', 'sub', 'mul', 'div'), 
                                       max_samples=0.9, verbose=1, 
                                       parsimony_coefficient=0.0004, 
                                       random_state=0, n_jobs=-1)
```
